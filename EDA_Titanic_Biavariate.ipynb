BIVARIATE ANALYSIS
Bivariate Analysis --> studying two variables to identify relationships or patterns.
1. Numerical-Numerical
2. Numerical-Categorical
3. Categorical-Categorical

In this example, we will explore how various features relate to the target variable: 'Survived'

```python
#Step 1: Survival Count by Gender (Do women survive more often than men?)
sns.countplot(data=df, x='Sex', hue='Survived', palette='Set2')
plt.title('Survival Count by Gender')
plt.xlabel('Sex')
plt.ylabel('Count')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()
```

```python
  # Step 2: Survival by Passenger Class (Was class a factor in survival?)
```

```python
sns.countplot(data=df, x='Pclass', hue='Survived', palette='Set1')
plt.title('Survival Count by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Count')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()
```

```python
# Survival Count by Age (Count Plot is not suitable)
sns.countplot(data=df, x='Age', hue='Survived', palette='Set1')
plt.title('Survival Count by Age')
plt.xlabel('Age')
plt.ylabel('Count')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()
```

```python
# Step 3: Survival vs Age (Boxplot)  Were children prioritized?
sns.boxplot(data=df, x='Survived', y='Age',palette='Set3')
plt.title('Age Distribution by Survival')
plt.xlabel('Survived')
plt.ylabel('Age')
plt.xticks([0, 1], ['No', 'Yes'])
plt.show()
```

```python
# Survival vs Fare (Boxplot)
# Did wealthy passengers survive more?
sns.boxplot(data=df, x='Survived', y='Fare', palette='coolwarm')
plt.title('Fare Distribution by Survival')
plt.xlabel('Survived')
plt.ylabel('Fare')
plt.xticks([0, 1], ['No', 'Yes'])
plt.show()
```

```python
# Survival by Embarkation Port (Did port of boarding affect survival?)  S = Southampton , C = Cherbourg ,Q = Queenstown
sns.countplot(data=df, x='Embarked', hue='Survived', palette='Set2')
plt.title('Survival Count by Embarkation Port')
plt.xlabel('Port of Embarkation')
plt.ylabel('Count')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()
```

```python
# Fare distribution class wiese
sns.barplot(data=df, x='Pclass', y='Fare', palette='coolwarm', hue='Sex')
plt.title('Fare Distribution by Class')
plt.xlabel('Class')
plt.ylabel('Fare')
plt.show()
```

```python
# Box Plot (Numerical-Categorical)
# Age-Gender
sns.boxplot(data=df, x='Sex', y='Age', palette='coolwarm')
plt.title('Relation between Age and Gender')
plt.xlabel('Sex')
plt.ylabel('Age')
plt.show()
```

```python
# Box Plot (Numerical-Categorical)
# Age-Gender
sns.boxplot(data=df, x='Sex', y='Age', palette='coolwarm',hue='Survived')
plt.title('Relation between Age and Gender')
plt.xlabel('Sex')
plt.ylabel('Age')
plt.show()
```
